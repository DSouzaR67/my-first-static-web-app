```python
# Función para extraer la calle y el código postal de una dirección
def extraer_calle_codigo_postal(direccion):
    partes = direccion.split(',')
    calle = partes[0].strip()
    codigo_postal = partes[-2].strip() if len(partes) > 1 else ""
    return calle, codigo_postal

# Extraer la primera ruta (leg)
leg = directions_result[0]['legs'][0]

# Extraer los pasos (steps) y sus duraciones, distancias, ubicaciones
steps = leg['steps']

# Lista para almacenar la información completa de cada paso con calle y código postal
informacion_pasos_con_calle_cp = []

# Iterar sobre cada paso para extraer la duración, distancia, ubicaciones, calle y código postal
for step in steps:
    step_duration = step['duration']['text']
    step_distance = step['distance']['text']
    step_start_location = step['start_location']
    step_end_location = step['end_location']
    step_start_address = step.get('start_address', leg['start_address'])
    step_end_address = step.get('end_address', leg['end_address'])
    calle_inicio, cp_inicio = extraer_calle_codigo_postal(step_start_address)
    calle_fin, cp_fin = extraer_calle_codigo_postal(step_end_address)
    informacion_pasos_con_calle_cp.append({
        'duracion': step_duration,
        'distancia': step_distance,
        'punto_inicio': step_start_location,
        'punto_fin': step_end_location,
        'start_address': step_start_address,
        'end_address': step_end_address,
        'calle_inicio': calle_inicio,
        'cp_inicio': cp_inicio,
        'calle_fin': calle_fin,
        'cp_fin': cp_fin
    })

# Imprimir la información completa de cada paso con calle y código postal
for i, info in enumerate(informacion_pasos_con_calle_cp, 1):
    print(f"Paso {i}:")
    print(f"  Duración: {info['duracion']}")
    print(f"  Distancia: {info['distancia']}")
    print(f"  Punto de Inicio: {info['punto_inicio']}")
    print(f"  Dirección de Inicio: {info['start_address']}")
    print(f"  Calle de Inicio: {info['calle_inicio']}, CP: {info['cp_inicio']}")
    print(f"  Punto Final: {info['punto_fin']}")
    print(f"  Dirección Final: {info['end_address']}")
    print(f"  Calle Final: {info['calle_fin']}, CP: {info['cp_fin']}")
    print("-" * 40)

# Información de la ruta completa
punto_inicio_ruta = leg['start_location']
punto_final_ruta = leg['end_location']
duracion_total_ruta = leg['duration']['text']
distancia_total_ruta = leg['distance']['text']
direccion_inicio_ruta = leg['start_address']
direccion_final_ruta = leg['end_address']

print("Información de la ruta completa:")
print(f"  Dirección de Inicio de la Ruta: {direccion_inicio_ruta}")
print(f"  Dirección Final de la Ruta: {direccion_final_ruta}")
print(f"  Punto de Inicio de la Ruta: {punto_inicio_ruta}")
print(f"  Punto Final de la Ruta: {punto_final_ruta}")
print(f"  Duración Total de la Ruta: {duracion_total_ruta}")
print(f"  Distancia Total de la Ruta: {distancia_total_ruta}")
```
